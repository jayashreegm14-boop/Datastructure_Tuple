```python
# Tuple is Used To Store Multivalue in Variable
#(1).Tuple Operator
tup1 = (10,20,30)
tup2 = (40,50)
tup3 = (10,20,30)
#Concatenation Operator
print("Addition:",tup1+tup2)
#Repitition Operator
print("Repetition:",tup1*2)
#Membership Operator
print("Membership:",20 in tup1)
#Not Membership Operator
print("Not Membership:",20 not in tup2)
#Identity Operator
print("Identity:",tup1 is tup3)
#Not Identity Operator
print("Not Identity:",tup1 is not tup3)
#Equality Operator
print("Equality:",tup1 == tup3)
#Inequality Operator
print("Inequality:",tup1 != tup2)
#(2).Tuple Operation
t = (1,2,3,4,5)
#Acess an Element
print("Acess:",t[1])
#Slicing an Element
print("Slicing:",t[2:4])
#Count an Element
print("Count:",t.count(1))
#Index an Element
print("Index:",t.index(2))
#(3).Built-in Function
lab = [10,20,30,40,50,60]
#Length of Tuple
print("Length:",len(lab))
#Maximum of Tuple
print("Maximum:",max(lab))
#Minimum of Tuple
print("Minimum:",min(lab))
#Sum of Tuple
print("Sum:",sum(lab))
#Average of Tuple
print("Average:",sum(lab)/len(lab))
```
